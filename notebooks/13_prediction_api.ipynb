## 3. Historical Feature Generation

Model membutuhkan historical features yang menggambarkan performa masing-masing tim sebelum pertandingan berlangsung.

Pada tahap training, fitur tersebut dibangun secara kronologis menggunakan riwayat pertandingan sebelumnya. Pada saat inference, prinsip yang sama harus dipertahankan: sistem hanya boleh menggunakan pertandingan yang telah terjadi sebelum waktu prediksi.

Untuk prototype API ini, historical dataset digunakan sebagai sumber riwayat pertandingan. Dari dataset tersebut, sistem akan mengambil pertandingan sebelumnya dari `home_team` maupun `away_team`, kemudian menghitung historical statistics yang dibutuhkan oleh model.

### 3.2 Buat Helper untuk Mengambil History Tim

```python
def get_team_history(team, data):
    home_matches = data[data["home_team"] == team]
    away_matches = data[data["away_team"] == team]

    return pd.concat(
        [home_matches, away_matches]
    ).sort_values("date")
```

### 3.3 Historical Matches Before Prediction

Pada saat melakukan prediksi, model hanya boleh menggunakan informasi yang tersedia sebelum pertandingan yang diprediksi. Oleh karena itu, riwayat pertandingan harus dibatasi berdasarkan tanggal prediksi.

Fungsi berikut mengambil seluruh pertandingan suatu tim yang berlangsung sebelum tanggal pertandingan yang akan diprediksi.

```python
def get_team_history_before(team, prediction_date, data):
    prediction_date = pd.to_datetime(prediction_date)

    team_matches = data[
        (data["home_team"] == team) |
        (data["away_team"] == team)
    ]

    return (
        team_matches[
            team_matches["date"] < prediction_date
        ]
        .sort_values("date")
        .reset_index(drop=True)
    )
```

### 3.4 Convert Matches to Team Perspective

Historical statistics dihitung dari sudut pandang masing-masing tim.

Karena sebuah tim dapat bermain sebagai home maupun away, informasi pertandingan perlu dikonversi terlebih dahulu menjadi:

- `result`
- `goals_scored`
- `goals_conceded`
- `goal_difference`

Dengan demikian, perhitungan historical features dapat menggunakan format yang konsisten terlepas dari posisi tim pada pertandingan tersebut.

```python
def build_team_records(team, matches):
    records = []

    for _, match in matches.iterrows():
        if match["home_team"] == team:
            goals_scored = match["home_score"]
            goals_conceded = match["away_score"]
        else:
            goals_scored = match["away_score"]
            goals_conceded = match["home_score"]

        if goals_scored > goals_conceded:
            result = "W"
        elif goals_scored < goals_conceded:
            result = "L"
        else:
            result = "D"

        records.append({
            "result": result,
            "goals_scored": goals_scored,
            "goals_conceded": goals_conceded,
            "goal_difference": goals_scored - goals_conceded
        })

    return records
```

### 3.5 Generate Historical Features

Setelah riwayat pertandingan dikonversi ke perspektif masing-masing tim, historical features dapat dihitung menggunakan statistik pertandingan sebelumnya.

Fitur yang dihasilkan harus konsisten dengan fitur yang digunakan pada tahap Historical Modeling dan Elo Modeling.

Untuk setiap tim, sistem menghitung:

- Win Rate 5 pertandingan terakhir
- Win Rate 10 pertandingan terakhir
- Rata-rata gol yang dicetak
- Rata-rata gol yang diterima
- Rata-rata goal difference

Proses ini dilakukan secara terpisah untuk home team dan away team.

```python
# WINRATE FUNCTION
import numpy as np


def calculate_winrate(history, last_n):
    """
    Calculate win rate from the last N matches.

    Parameters
    ----------
    history : list
        Match history of a team.
    last_n : int
        Number of recent matches to consider.

    Returns
    -------
    float
        Win rate between 0 and 1.
    """
    
    if len(history) == 0:
        return np.nan

    recent_matches = history[-last_n:]

    wins = sum(
        match["result"] == "W"
        for match in recent_matches
    )

    return wins / len(recent_matches)


# AVG SCORE FUNCTION
def calculate_avg_scored(history, last_n):
    if len(history) == 0:
        return np.nan

    recent_matches = history[-last_n:]

    return np.mean([
        match["goals_scored"]
        for match in recent_matches
    ])


# AVG CONCEDED FUNCTION
def calculate_avg_conceded(history, last_n):
    if len(history) == 0:
        return np.nan

    recent_matches = history[-last_n:]

    return np.mean([
        match["goals_conceded"]
        for match in recent_matches
    ])

# AVG GOAL DIFFERENCE FUNCTION
def calculate_goal_difference(history, last_n):
    if len(history) == 0:
        return np.nan

    recent_matches = history[-last_n:]

    return np.mean([
        match["goal_difference"]
        for match in recent_matches
    ])
```

```python
def generate_team_historical_features(team, prediction_date, data):
    history = get_team_history_before(
        team,
        prediction_date,
        data
    )

    records = build_team_records(
        team,
        history
    )

    features = {
        "last5_winrate": calculate_winrate(records, 5),
        "last10_winrate": calculate_winrate(records, 10),
        "avg_goals_scored": calculate_avg_scored(records, 5),
        "avg_goals_conceded": calculate_avg_conceded(records, 5),
        "goal_difference_form": calculate_goal_difference(records, 5)
    }

    return features
```

### 3.6 Generate Features for a Match

```python
def generate_historical_features(
    home_team,
    away_team,
    prediction_date,
    data
):
    home_features = generate_team_historical_features(
        home_team,
        prediction_date,
        data
    )

    away_features = generate_team_historical_features(
        away_team,
        prediction_date,
        data
    )

    return {
        "home_last5_winrate": home_features["last5_winrate"],
        "home_last10_winrate": home_features["last10_winrate"],
        "home_avg_goals_scored": home_features["avg_goals_scored"],
        "home_avg_goals_conceded": home_features["avg_goals_conceded"],
        "home_goal_difference_form": home_features["goal_difference_form"],

        "away_last5_winrate": away_features["last5_winrate"],
        "away_last10_winrate": away_features["last10_winrate"],
        "away_avg_goals_scored": away_features["avg_goals_scored"],
        "away_avg_goals_conceded": away_features["avg_goals_conceded"],
        "away_goal_difference_form": away_features["goal_difference_form"]
    }
```

## 4. Elo Generation

### 4.1 Load Elo Dataset

```python
# Load Elo feature dataset

elo_data = pd.read_csv(
    "../data/processed/elo_features.csv",
    parse_dates=["date"]
)

elo_data = elo_data.sort_values("date").reset_index(drop=True)

elo_data.head()
```

### 4.2 Fungsi Untuk Mengambil Elo Terakhir Sebuah Tim

```python
def get_team_elo(team, prediction_date, data):
    """
    Get the latest Elo rating of a team before prediction_date.
    """

    prediction_date = pd.to_datetime(prediction_date)

    past_matches = data[data["date"] < prediction_date]

    if past_matches.empty:
        return np.nan

    home_matches = past_matches[
        past_matches["home_team"] == team
    ]

    away_matches = past_matches[
        past_matches["away_team"] == team
    ]

    if home_matches.empty and away_matches.empty:
        return np.nan

    # Latest match as home
    latest_home = (
        home_matches.iloc[-1]
        if not home_matches.empty
        else None
    )

    # Latest match as away
    latest_away = (
        away_matches.iloc[-1]
        if not away_matches.empty
        else None
    )

    # Select the chronologically latest match
    if latest_home is None:
        return latest_away["away_elo_after"]

    if latest_away is None:
        return latest_home["home_elo_after"]

    if latest_home["date"] > latest_away["date"]:
        return latest_home["home_elo_after"]

    return latest_away["away_elo_after"]
```

### 4.3 Generate Elo Features

```python
def generate_elo_features(
    home_team,
    away_team,
    prediction_date,
    data
):
    """
    Generate Elo features for a future match.
    """

    home_elo = get_team_elo(
        home_team,
        prediction_date,
        data
    )

    away_elo = get_team_elo(
        away_team,
        prediction_date,
        data
    )

    features = {
        "home_elo_before": home_elo,
        "away_elo_before": away_elo
    }

    return features
```

## 5. Feature Assembly

Tujuan: Menggabungkan semua informasi yang dibutuhkan model final menjadi satu DataFrame dengan schema yang sama persis seperti saat training.

### 5.1 Baseline Feature Generator

```python
def generate_baseline_features(
    home_team,
    away_team,
    tournament,
    neutral
):
    """
    Generate baseline features for a future match.
    """

    features = {
        "home_team": home_team,
        "away_team": away_team,
        "tournament": tournament,
        "neutral": neutral
    }

    return features
```

```python
test_baseline_features = generate_baseline_features(
    home_team="Brazil",
    away_team="Argentina",
    tournament="Friendly",
    neutral=False
)

test_baseline_features
```

### 5.2 Gabungkan Semua Feature

```python
def generate_prediction_features(
    home_team,
    away_team,
    tournament,
    neutral,
    prediction_date,
    historical_data,
    elo_data
):
    """
    Generate the complete feature set required by the
    final Logistic Regression + Elo model.
    """

    baseline_features = generate_baseline_features(
        home_team,
        away_team,
        tournament,
        neutral
    )

    historical_features = generate_historical_features(
        home_team,
        away_team,
        prediction_date,
        historical_data
    )

    elo_features = generate_elo_features(
        home_team,
        away_team,
        prediction_date,
        elo_data
    )

    features = {
        **baseline_features,
        **historical_features,
        **elo_features
    }

    return pd.DataFrame([features])
```

## 7. Prediction Function

### 7.1 Buat `predict_match()`

```python
RESULT_LABELS = {
    "H": "Home Win",
    "D": "Draw",
    "A": "Away Win"
}
```

```python
def predict_match(
    home_team,
    away_team,
    tournament,
    neutral,
    prediction_date,
    historical_data,
    elo_data,
    model
):
    """
    Generate features and predict match outcome.
    """

    features = generate_prediction_features(
        home_team=home_team,
        away_team=away_team,
        tournament=tournament,
        neutral=neutral,
        prediction_date=prediction_date,
        historical_data=historical_data,
        elo_data=elo_data
    )

    prediction = model.predict(features)[0]

    probabilities = model.predict_proba(features)[0]

    probability_dict = dict(
        zip(
            model.classes_,
            probabilities
        )
    )

    return {
        "home_team": home_team,
        "away_team": away_team,
        "prediction": prediction,
        "prediction_label": RESULT_LABELS[prediction],
        "probabilities": probability_dict
    }
```

### 7.2 Test End-to-End
